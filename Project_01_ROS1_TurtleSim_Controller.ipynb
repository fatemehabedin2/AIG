# Project 1 — TurtleSim Controller

````python
#!/usr/bin/env python

import rospy
import sys
import select
import termios
import tty
from geometry_msgs.msg import Twist


def get_key(timeout):
    settings = termios.tcgetattr(sys.stdin)
    tty.setraw(sys.stdin.fileno())

    rlist, _, _ = select.select([sys.stdin], [], [], timeout)

    if rlist:
        key = sys.stdin.read(1)
    else:
        key = ""

    termios.tcsetattr(sys.stdin, termios.TCSADRAIN, settings)
    return key


def key_to_twist(key):
    msg = Twist()

    if key == "w":
        msg.linear.x = 2.0
    elif key == "s":
        msg.linear.x = -2.0
    elif key == "a":
        msg.angular.z = 2.0
    elif key == "d":
        msg.angular.z = -2.0

    return msg


def main():
    rospy.init_node("turtle_keyboard_controller")

    if len(sys.argv) < 2:
        rospy.logerr("Usage: rosrun proj1_turtlesim turtle_controller.py turtle1")
        return

    turtle_name = sys.argv[1]
    topic_name = "/" + turtle_name + "/cmd_vel"

    rospy.loginfo("Controlling turtle: %s", turtle_name)
    rospy.loginfo("Publishing velocity commands to: %s", topic_name)

    pub = rospy.Publisher(topic_name, Twist, queue_size=10)

    print("Controls:")
    print("w = forward")
    print("s = backward")
    print("a = turn left")
    print("d = turn right")
    print("q = quit")
    print("No key / invalid key = stop")

    rate = rospy.Rate(10)

    while not rospy.is_shutdown():
        key = get_key(0.1)

        if key == "q":
            pub.publish(Twist())
            rospy.loginfo("Stopping %s and shutting down controller.", turtle_name)
            break

        msg = key_to_twist(key)

        # If key is invalid or no key is pressed, key_to_twist returns zero velocity.
        pub.publish(msg)

        rate.sleep()


if __name__ == "__main__":
    main()
````

## Project goal
Build a ROS package that controls a TurtleSim turtle using keyboard input. Your controller should accept the turtle name as an argument and publish velocity commands to the correct topic.

## Required behavior
- Package name: `proj1_turtlesim`.
- The node accepts a target turtle name, for example `turtle1` or `turtle2`.
- The controller publishes `geometry_msgs/Twist` messages.
- `w/a/s/d` provide basic motion; diagonal/combined options may be implemented with `q/e/z/c` or true multi-key input.
- The turtle stops when no valid key is pressed.


## Suggested setup
```bash
mkdir -p ~/proj1_ws/src
cd ~/proj1_ws/src
catkin_create_pkg proj1_turtlesim rospy geometry_msgs
mkdir -p proj1_turtlesim/scripts
cd ~/proj1_ws
catkin_make
source devel/setup.bash
```


## Implementation guidance
Structure your node into functions:
1. Parse the target turtle name from `sys.argv`.
2. Build the topic name: `/{turtle_name}/cmd_vel`.
3. Read keys without blocking the ROS shutdown process.
4. Convert keys to linear/angular velocity.
5. Publish zero velocity when stopping.
6. Log which turtle is being controlled.


## Testing protocol



## Assessment questions
1. What command did you use to create the package?
catkin_create_pkg proj1_turtlesim rospy geometry_msgs

2. Which ROS message type did your node publish, and why?
The node published the geometry_msgs/Twist message type.
I used Twist because the turtlesim node expects velocity commands on the /cmd_vel topic in the form of linear and angular velocities. The Twist message allows the controller to move the turtle forward and backward using linear.x and rotate it left and right using angular.z, which maps directly to the w, a, s, and d keyboard controls.

3. How do you launch ROS, TurtleSim, and your controller together?
I launched the system using separate terminals. First, I started the ROS master with roscore. Next, I launched TurtleSim using rosrun turtlesim turtlesim_node. Then, after sourcing my workspace with source ~/proj1_ws/devel/setup.bash, I started my controller node using rosrun proj1_turtlesim turtle_controller.py turtle1. I also spawned a second turtle with rosservice call /spawn 5 5 0 turtle2 and launched a second instance of the controller using rosrun proj1_turtlesim turtle_controller.py turtle2. This allowed each controller to publish velocity commands to its respective turtle.

4. How did you verify that your node publishes correctly?
I verified that my node was publishing correctly using both rostopic echo and rqt_graph. I ran rostopic echo /turtle1/cmd_vel and rostopic echo /turtle2/cmd_vel to observe the geometry_msgs/Twist messages being published when I pressed the keyboard controls. I also used rqt_graph to confirm that my turtle_keyboard_controller node was connected to the appropriate /cmd_vel topics and that the TurtleSim node was subscribing to those messages. Finally, I confirmed correct behavior by observing the turtles move in response to the published velocity commands.
